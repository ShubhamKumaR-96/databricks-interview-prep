--     • CREATE TABLE ... SHALLOW CLONE copies only metadata; data files are shared
--     • Inserted new records into shallow clone
--     • Verified that changes in shallow clone do NOT affect the source table

-- ════════════════════════════════════════════════════
-- 2️⃣ SPECIAL / KEY POINTS
-- ════════════════════════════════════════════════════

--  🔹 Time Travel Syntax Options
--     • Version-based : SELECT * FROM table@V0
--     • Timestamp-based: SELECT * FROM table@2024-01-01_12:00:00
--     • Used for auditing, rollback, and comparing historical states

--  🔹 Deep Clone vs Shallow Clone
--     ┌──────────────┬──────────────────────┬────────────────────────┐
--     │   Feature    │     Deep Clone       │     Shallow Clone      │
--     ├──────────────┼──────────────────────┼────────────────────────┤
--     │ Data Copy    │ Full copy of data   │ Only metadata copied   │
--     │ Storage Cost │ High (duplicates)   │ Low (shares data files) │
--     │ Independence │ Fully independent   │ Shares source data files│
--     │ Best For     │ Dev/test, backup    │ Quick snapshot, testing │
--     └──────────────┴──────────────────────┴────────────────────────┘

--  🔹 DESCRIBE HISTORY
--     • Shows version, timestamp, operation (WRITE, INSERT), operation metrics
--     • Each INSERT/UPDATE/DELETE/MERGE creates a new version
--     • Essential for auditing and time travel

--  🔹 Temporary View
--     • Lives only for the session — not stored in catalog
--     • Useful for intermediate transformations and ad-hoc analysis
--     • Cannot be shared across sessions or users

-- ════════════════════════════════════════════════════
-- 3️⃣ MOST ASKED INTERVIEW QUESTIONS
-- ════════════════════════════════════════════════════

--  Q1. What is Delta Lake and how is it different from Parquet?
--      → Delta adds ACID transactions, schema enforcement, time travel,
--        and upsert/delete support on top of Parquet files.

--  Q2. What is Time Travel in Delta Lake? How do you use it?
--      → Querying previous versions of a table using @V<n> or @timestamp.
--      → e.g., SELECT * FROM table@V0
--      → Powered by the Delta transaction log (_delta_log).

--  Q3. What is the difference between Deep Clone and Shallow Clone?
--      → Deep Clone: copies both data and metadata → fully independent table.
--      → Shallow Clone: copies only metadata → shares underlying data files.
--      → Shallow clones must be on same cluster/workspace as source.

--  Q4. What does DESCRIBE HISTORY return?
--      → A log of all operations (INSERT, UPDATE, DELETE, MERGE, etc.)
--        with version number, timestamp, operation type, and metrics.

--  Q5. What is the difference between a Temporary View and a regular table?
--      → Temp View: session-scoped, not persisted, not in catalog.
--      → Table: persistent, stored in catalog, survives session end.

--  Q6. Does inserting into a Deep Clone affect the source table?
--      → No. Deep clone is a fully independent copy.

--  Q7. Does inserting into a Shallow Clone affect the source table?
--      → No. New writes go to new files in the clone. Source data files
--        remain unchanged (copy-on-write semantics).

--  Q8. How many versions are retained for time travel by default?
--      → Controlled by table properties:
--      → delta.logRetentionDuration (default 30 days)
--      → delta.deletedFileRetentionDuration (default 7 days)

--  Q9. What is DESCRIBE EXTENDED used for?
--      → Shows comprehensive table metadata: schema, partition columns,
--        table location, provider, and table properties.

--  Q10. Can you clone a view or a non-Delta table?
--      → No. CLONE works only on Delta tables.
